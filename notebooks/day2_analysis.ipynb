```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
data = 
{
    "day": [1, 2, 3, 4, 5],
    "tasks_completed": [3, 5, 4, 7, 6]
}

df = pd.DataFrame(data)

df
```

```python
plt.plot(df["day"], df["tasks_completed"], marker="o")
plt.xlabel("Day")
plt.ylabel("Tasks Completed")
plt.title("Tasks Completed Per Day")
plt.show()
```

```python
fig = px.line
(
    df,
    x="day",
    y="tasks_completed",
    title="Tasks Completed Per Day"
)

fig.show()
```
